### Cleaning Data

#### Fixing df_day

```python
df_day['yr'] = df_day['yr'].map({0: '2011', 1: '2012'})
df_day['season'] = df_day['season'].map({1: 'Spring', 2: 'Summer', 3: 'Fall', 4: 'Winter'})
```

```python
df_day.head()
```

#### Fixing df_hour

```python
df_hour['yr'] = df_hour['yr'].map({0: '2011', 1: '2012'})
df_hour['season'] = df_hour['season'].map({1: 'Spring', 2: 'Summer', 3: 'Fall', 4: 'Winter'})
```

```python
df_hour.head()
```

**Insight:** (Opsional)
- Membersihkan data merupakan tahap krusial untuk memastikan kelancaran dalam pengolahan data serta mempermudah interpretasi data sebagai suatu insight.
- Pada cleaning data, saya hanya melakukan perubahan tipe data season dan yr dari numerik menjadi string untuk mempermudah pembacaan.

## Exploratory Data Analysis (EDA)

### Explore df_day

```python
df_day.sample(5)
```

```python
df_day.describe(include="all")
```

```python
pertumbuhan = df_day.groupby(by=["yr", "season"]).agg({
    "cnt": "sum"
}).reset_index()

display(pertumbuhan)
```

### Explore df_hour

```python
df_hour.sample(15)
```

```python
df_hour.describe(include="all")
```

```python
permintaan = df_hour.groupby(by='hr').agg({
    "cnt": "mean"
}).reset_index().sort_values(by='cnt', ascending=False)

display(permintaan)
```

**Insight:** (Opsional)
- Exploratory Data Analysis digunakan untuk menjelajahi data agar kita bisa menemukan jawaban kita. Dalam penjelajahan ini, data tidak hanya sekedar dibaca, namun juga dimanipulasi baik digabungkan atau diurutkan
- Pada bagian ini, saya memanipulasi setiap data dan mengurutkannya. Untuk df_day, saya mengurutkan lalu menjumlahkan kolom count untuk mendapatkan total dari cnt (jumlah) pada tahun dan musim itu. Untuk df_hour, saya mengurutkan lalu merata-ratakan kolom count untuk mendapatkan rata-rata pengguna berdasarkan jam tertentu, lalu mengurutkannya berdasarkan cnt terbanyak

## Visualization & Explanatory Analysis

### Pertanyaan 1: Apakah ada pertumbuhan jumlah penyewa sepeda dari tahun 2011 sampai 2012 berdasarkan musim-musim?

```python
plt.figure(figsize=(10, 6))
sns.barplot(data=pertumbuhan, x='season', y='cnt', hue='yr', palette='viridis')
plt.title('Total Bike Rentals by Season (2011 vs 2012)')
plt.xlabel('Season')
plt.ylabel('Total Rentals')
plt.show()
```

### Pertanyaan 2: Pada jam berapa demand penyewaan sepeda mencapai titik puncakanya?

```python
plt.figure(figsize=(12, 6))
sns.barplot(data=permintaan, x='hr', y='cnt', palette='coolwarm')
plt.title('Average Bike Rental Demand by Hour')
plt.xlabel('Hour of the Day')
plt.ylabel('Average Total Rentals')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
