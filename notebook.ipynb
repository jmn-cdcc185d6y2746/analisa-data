## Exploratory Data Analysis (EDA)

```python
pertumbuhan = df_day.groupby(by=["yr", "season"]).agg({
    "cnt": "sum"
}).reset_index()

display(pertumbuhan)
```

```python
df_hour.sample(15)
```

```python
df_hour.describe(include="all")
```

```python
permintaan = df_hour.groupby(by='hr').agg({
    "cnt": "mean"
}).reset_index().sort_values(by='cnt', ascending=False)

display(permintaan)
```

Untuk df_hour, saya mengurutkan lalu merata-ratakan kolom count untuk mendapatkan rata-rata pengguna berdasarkan jam tertentu, lalu mengurutkannya berdasarkan cnt terbanyak

## Visualization & Explanatory Analysis

### Pertanyaan 1: Apakah ada pertumbuhan jumlah penyewa sepeda dari tahun 2011 sampai 2012 berdasarkan musim-musim?

```python
plt.figure(figsize=(10, 6))
sns.barplot(data=pertumbuhan, x='season', y='cnt', hue='yr')
plt.title('Total Bike Rentals by Season (2011 vs 2012)')
plt.xlabel('Season')
plt.ylabel('Total Rentals')
plt.show()
```

### Pertanyaan 2: Pada jam berapa demand penyewaan sepeda mencapai titik puncakanya dalam suatu hari?

```python
plt.figure(figsize=(12, 6))
sns.lineplot(data=permintaan, x='hr', y='cnt')
plt.title('Average Bike Rental Demand by Hour')
plt.xlabel('Hour of the Day')
plt.ylabel('Average Total Rentals')
plt.xticks(range(0, 24), rotation=45) # Set x-ticks for all hours (0-23)

# Get the peak hour from the 'permintaan' DataFrame
peak_hour = permintaan.loc[permintaan['cnt'].idxmax()]['hr']
peak_demand = permintaan['cnt'].max()

# Add a vertical line at the peak hour
plt.axvline(x=peak_hour, color='red', linestyle='--', label=f'Puncak: {int(peak_hour)}:00')
plt.legend()

plt.tight_layout()
plt.show()
```

**Insight:** (Opsional)
- Dari pertanyaan pertama, dapat dilihat bahwa ada perkembangan signifikan dari tahun 2011 ke 2012. Sesuai dengan trendnya juga, total penyewaan tertinggi terdapat pada musim gugur
- Dari pertanyaan kedua, dapat dilihat bahwa penyewaan paling banyak memuncak di jam ke-17. Ini mengindikasikan bahwa sebagian besar penyewa menggunakan sepeda untuk pulang dari kegiatan masing-masing
